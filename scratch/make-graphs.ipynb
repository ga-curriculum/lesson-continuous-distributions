```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

```python
x = stats.norm(0, 1)
```

```python
smps = x.rvs(100)
```

```python
plt.figure(figsize=(8, 6))
plt.hist(smps, color='#e51b24', edgecolor='white');
plt.xticks([])
plt.yticks([])

plt.savefig('disc-dist.png', dpi=192)
```

```python
f = x.pdf
vals = np.linspace(-3, 3, 1000)
```

```python
plt.figure(figsize=(8, 6))
plt.plot(vals, f(vals), color='#e51b24', linewidth=3)
plt.fill_between(vals, 0, f(vals), color='#e51b24', alpha=0.3)
plt.xticks([])
plt.yticks([])

plt.savefig('cts-dist.png', dpi=192)
```

```python
vals_cut = vals[vals < 1]

plt.figure(figsize=(8, 6))
plt.plot(vals, f(vals), color='#e51b24', linewidth=3)
plt.fill_between(vals_cut, 0, f(vals_cut), color='#e51b24', alpha=0.3)
plt.xticks([1], labels=['$x$'], size=20)
plt.yticks([])

plt.savefig('cdf.png', dpi=192)
```

```python
vals_cut = vals[(vals > -1/2) & (vals < 1)]

plt.figure(figsize=(8, 6))
plt.plot(vals, f(vals), color='#e51b24', linewidth=3)
plt.fill_between(vals_cut, 0, f(vals_cut), color='#e51b24', alpha=0.3)
plt.xticks([-1/2, 1], labels=['$a$', '$b$'], size=20)
plt.yticks([])

plt.savefig('probability.png', dpi=192)
```
